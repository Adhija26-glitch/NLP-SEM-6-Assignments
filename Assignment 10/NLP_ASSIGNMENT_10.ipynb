```python
!pip install -q nltk

import nltk
nltk.download('punkt')

from nltk.util import ngrams
from nltk.tokenize import word_tokenize
from collections import defaultdict, Counter

text = """natural language processing is very useful and natural language processing is widely used in many applications
machine learning is a part of artificial intelligence and natural language processing is a part of it"""

tokens = word_tokenize(text.lower())

bigrams = list(ngrams(tokens, 2))

model = defaultdict(Counter)

for w1, w2 in bigrams:
    model[w1][w2] += 1

def predict_next_word(word):
    if word in model:
        return model[word].most_common(3)
    else:
        return []

inputs = ["natural", "language", "processing", "machine", "artificial"]

for word in inputs:
    print("Input:", word)
    print("Predictions:", predict_next_word(word))
    print()
```
