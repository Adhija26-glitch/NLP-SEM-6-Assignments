```python
!pip install -q nltk

import nltk
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.wsd import lesk
from nltk.tokenize import word_tokenize

sentences = [
    "I went to the bank to deposit money.",
    "The river bank was full of water.",
    "He is playing cricket with a bat.",
    "The bat flew out of the cave."
]

target_words = ["bank", "bank", "bat", "bat"]

for sentence, word in zip(sentences, target_words):
    tokens = word_tokenize(sentence)
    sense = lesk(tokens, word)

    print("Sentence:", sentence)
    print("Word:", word)

    if sense:
        print("Sense:", sense.name())
        print("Definition:", sense.definition())
        print("Examples:", sense.examples())
    else:
        print("Sense: Not found")

    print("-" * 50)
```
