```python
!pip install -q transformers sentencepiece

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "Helsinki-NLP/opus-mt-en-hi"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

# Function to perform translation directly without pipeline
def translate_text(text):
    inputs = tokenizer(text, return_tensors="pt", padding=True, truncation=True)
    translated_tokens = model.generate(**inputs)
    return tokenizer.decode(translated_tokens[0], skip_special_tokens=True)

text = "The government has launched a new healthcare initiative."

result = translate_text(text)

print("English :", text)
print("Hindi   :", result)

print()

texts = [
    "India is improving its digital infrastructure.",
    "Public health is an important concern.",
    "Education helps in national development.",
    "The government is working on new policies for economic growth."
]

for t in texts:
    translated = translate_text(t)
    print("English :", t)
    print("Hindi   :", translated)
    print()
```
